```python
from config import GeminiConfig


def extract_resume_text(file_path:str) -> str:
    """Extract the resume"""
    try:
        client = genai.Client(api_key=GeminiConfig.get_gemini_api_key())
        uploaded_file = client.files.upload(file=file_path)


        total_token_inputs = client.models.count_tokens(
            model="gemini-2.5-flash", contents=uploaded_file
        )
        print(total_token_inputs)

        response = client.models.generate_content(
            model="gemini-2.5-flash",
            contents=[uploaded_file, "Extract all text exactly as it appears:"]
        )
        print("===================",response)

        print("First when uploading Total input tokens: %s", total_token_inputs)

        if hasattr(response, 'text'):
                return response.text
        if hasattr(response, 'candidates') and response.candidates:
            return response.candidates[0].content.parts[0].text
    except Exception as e:
                return e
```

```python
import json
from gemini_client import GeminiClient
import google.generativeai as genaiA
from helpers import extract_and_validate_json
from datetime import date

# Define the few-shot example data directly in the script
# Example 1: Software Engineer
example_resume_text_1 = """
John Doe
(555) 123-4567 | john.doe@email.com | linkedin.com/in/johndoe

Summary
Full-stack software engineer with 7+ years of experience building scalable web applications. Expert in modern JavaScript frameworks and backend systems with a passion for clean code and robust architecture.

Skills
JavaScript, React, Node.js, Express.js, MongoDB, PostgreSQL, Docker, AWS, Git, RESTful APIs

Experience
Lead Software Engineer | Web Solutions Inc. | Jan 2021 – Present
- Led a team of 5 to develop and deploy a new e-commerce platform using the MERN stack.
- Designed and implemented a GraphQL API for data fetching, reducing client-side requests by 60%.
- Mentored junior developers on best practices for React component design and state management.

Senior Developer | Tech Innovations | Jun 2018 – Dec 2020
- Built and maintained the company’s primary marketing website with React and Next.js.
- Integrated third-party payment gateways and user authentication systems.
- Wrote unit and integration tests using Jest and Enzyme to ensure code quality.

Education
B.S. in Computer Science | University of California, Berkeley | 2018

Projects
- Real-time Chat App: A full-stack chat application built with React, Node.js, and Socket.IO.
- Personal Blog: A static site generated with Gatsby.js and hosted on Netlify.
"""

example_json_output_1 = """
{
  "first_name": "John",
  "last_name": "Doe",
  "name": "John Doe",
  "email": "john.doe@email.com",
  "linkedin_url": "https://www.linkedin.com/in/johndoe",
  "phone": "(555) 123-4567",
  "summary": "Full-stack software engineer with 7+ years of experience building scalable web applications. Expert in modern JavaScript frameworks and backend systems with a passion for clean code and robust architecture.",
  "skills": [
    {
      "name": "JavaScript",
      "rating": 10,
      "description": "Expert-level proficiency demonstrated by extensive use in multiple roles, leadership in development, and mention of advanced frameworks and libraries. Mentored junior developers, implying authoritative knowledge.",
      "sub_skills": [
        "Web Development",
        "Frontend",
        "Backend",
        "ES6+",
        "Async/Await"
      ]
    },
    {
      "name": "React",
      "rating": 9,
      "description": "Advanced proficiency shown by leading development of major platforms and mentoring on best practices. Mention of specific libraries like Next.js and Gatsby.js further justifies a high rating.",
      "sub_skills": [
        "React.js",
        "Next.js",
        "Gatsby.js",
        "Redux",
        "State Management"
      ]
    },
    {
      "name": "Node.js",
      "rating": 8,
      "description": "Advanced proficiency demonstrated by building a full-stack platform and a real-time chat application. Mention of Express.js and Socket.IO provides concrete evidence of use.",
      "sub_skills": [
        "Express.js",
        "Backend Development",
        "APIs",
        "Socket.IO"
      ]
    },
    {
      "name": "MongoDB",
      "rating": 7,
      "description": "Used in the MERN stack for a major e-commerce platform. While extensive details aren't given, its role in a core project indicates a solid understanding of its use.",
      "sub_skills": [
        "NoSQL",
        "Database Design",
        "Mongoose"
      ]
    },
    {
      "name": "AWS",
      "rating": 6,
      "description": "Mentioned as a general skill, but the resume lacks specific details on which services were used or how they were applied. The rating is intermediate, assuming some practical use in deployment.",
      "sub_skills": [
        "Cloud Computing",
        "Deployment"
      ]
    },
    {
      "name": "Git",
      "rating": 8,
      "description": "Standard tool for a senior developer. Implied frequent and complex use, including collaboration with a team of 5, which warrants an advanced rating.",
      "sub_skills": [
        "Version Control",
        "GitHub",
        "GitFlow"
      ]
    }
  ],
  "experience": [
    {
      "title": "Lead Software Engineer",
      "company": "Web Solutions Inc.",
      "duration": "Jan 2021 – Present",
      "description": "Led a team of 5 to develop and deploy a new e-commerce platform using the MERN stack. Designed and implemented a GraphQL API for data fetching, reducing client-side requests by 60%. Mentored junior developers on best practices for React component design and state management."
    },
    {
      "title": "Senior Developer",
      "company": "Tech Innovations",
      "duration": "Jun 2018 – Dec 2020",
      "description": "Built and maintained the company’s primary marketing website with React and Next.js. Integrated third-party payment gateways and user authentication systems. Wrote unit and integration tests using Jest and Enzyme to ensure code quality."
    }
  ],
  "total_experience": {
    "years": "7 years 9 months",
    "description": "John Doe has a total of 7 years and 9 months of professional experience in software engineering, primarily focused on full-stack web development with JavaScript, React, and Node.js. He has held senior and lead roles, demonstrating leadership and technical expertise in building and deploying scalable web applications."
  },
  "education": [
    {
      "degree": "B.S. in Computer Science",
      "institution": "University of California, Berkeley",
      "year": "2018"
    }
  ],
  "certifications": [],
  "other_qualifications": "John's academic background includes a Bachelor of Science in Computer Science from the University of California, Berkeley.",
  "projects": [
    {
      "name": "Real-time Chat App",
      "description": "A full-stack chat application built with React, Node.js, and Socket.IO."
    },
    {
      "name": "Personal Blog",
      "description": "A static site generated with Gatsby.js and hosted on Netlify."
    }
  ],
  "is_valid_resume": true
}
"""

# Example 2: Data Scientist
example_resume_text_2 = """
Jane Smith
(555) 987-6543 | jane.smith@email.com | linkedin.com/in/janesmith

Summary
Data scientist with 4 years of experience specializing in Natural Language Processing (NLP) and machine learning. Proficient in Python, with a strong background in developing and deploying models for text analysis and information retrieval.

Skills
Python, NLP, Machine Learning, PyTorch, Hugging Face, spaCy, Pandas, NumPy, Scikit-learn, SQL, AWS S3

Experience
Data Scientist | Innovate AI Labs | Feb 2022 – Present
- Led the development of an NLP pipeline for sentiment analysis of customer reviews, achieving 92% accuracy.
- Fine-tuned large language models (LLMs) from the Hugging Face library for specific domain tasks.
- Utilized AWS S3 for data storage and management.

Junior Data Scientist | Quant Analytics Corp | Aug 2020 – Feb 2022
- Preprocessed and cleaned large-scale text datasets for various machine learning projects.
- Built and trained predictive models for customer churn analysis using Scikit-learn.
- Collaborated with engineering teams to deploy models into production environments.

Education
M.S. in Data Science | Stanford University | 2020
B.A. in Statistics | Harvard University | 2018

Certifications
- Google Cloud Certified Professional Data Engineer

Projects
- Prompt Engineering Tool: A Python application that optimizes prompts for the Gemini API using few-shot and chain-of-thought techniques.
- Text Summarizer: A model built with PyTorch and transformers to generate concise summaries of articles.
"""

example_json_output_2 = """
{
  "first_name": "Jane",
  "last_name": "Smith",
  "name": "Jane Smith",
  "email": "jane.smith@email.com",
  "linkedin_url": "https://www.linkedin.com/in/janesmith",
  "phone": "(555) 987-6543",
  "summary": "Data scientist with 4 years of experience specializing in Natural Language Processing (NLP) and machine learning. Proficient in Python, with a strong background in developing and deploying models for text analysis and information retrieval.",
  "skills": [
    {
      "name": "Python",
      "rating": 9,
      "description": "High-level proficiency demonstrated by leading development of complex NLP pipelines and building multiple models. Used in conjunction with several advanced libraries like PyTorch, Hugging Face, and spaCy.",
      "sub_skills": [
        "Data Science",
        "Scripting",
        "Machine Learning",
        "Data Analysis"
      ]
    },
    {
      "name": "NLP",
      "rating": 10,
      "description": "Expert-level proficiency demonstrated by leading the development of a sentiment analysis pipeline and fine-tuning LLMs. The project section on 'Prompt Engineering Tool' also highlights deep, specialized knowledge in a very relevant and current subfield.",
      "sub_skills": [
        "LLMs",
        "Prompt Engineering",
        "Text Analysis",
        "Information Retrieval",
        "Sentiment Analysis",
        "Tokenization"
      ]
    },
    {
      "name": "Machine Learning",
      "rating": 9,
      "description": "Extensive experience building and deploying various models, including for sentiment analysis and customer churn. Mentions use of specific libraries like Scikit-learn and PyTorch.",
      "sub_skills": [
        "Supervised Learning",
        "Unsupervised Learning",
        "Model Deployment",
        "Model Training"
      ]
    },
    {
      "name": "PyTorch",
      "rating": 8,
      "description": "Used for building a text summarizer project and fine-tuning models. This demonstrates solid hands-on experience with the framework.",
      "sub_skills": [
        "Deep Learning",
        "Neural Networks",
        "Transformers"
      ]
    },
    {
      "name": "Hugging Face",
      "rating": 8,
      "description": "Specifically mentioned for fine-tuning large language models, indicating hands-on, advanced use of this critical NLP library.",
      "sub_skills": [
        "Transformers",
        "LLMs",
        "Model Fine-tuning"
      ]
    },
    {
      "name": "AWS S3",
      "rating": 6,
      "description": "Mentioned for data storage. While it's a key cloud service, the description doesn't detail complex use cases, justifying an intermediate rating.",
      "sub_skills": [
        "Cloud Storage",
        "Data Management"
      ]
    }
  ],
  "experience": [
    {
      "title": "Data Scientist",
      "company": "Innovate AI Labs",
      "duration": "Feb 2022 – Present",
      "description": "Led the development of an NLP pipeline for sentiment analysis of customer reviews, achieving 92% accuracy. Fine-tuned large language models (LLMs) from the Hugging Face library for specific domain tasks. Utilized AWS S3 for data storage and management."
    },
    {
      "title": "Junior Data Scientist",
      "company": "Quant Analytics Corp",
      "duration": "Aug 2020 – Feb 2022",
      "description": "Preprocessed and cleaned large-scale text datasets for various machine learning projects. Built and trained predictive models for customer churn analysis using Scikit-learn. Collaborated with engineering teams to deploy models into production environments."
    }
  ],
  "total_experience": {
    "years": "4 years 1 months",
    "description": "Jane Smith has accumulated 4 years and 1 month of professional experience in data science, with a strong specialization in Natural Language Processing and machine learning. She has worked on projects involving sentiment analysis, LLM fine-tuning, and model deployment."
  },
  "education": [
    {
      "degree": "M.S. in Data Science",
      "institution": "Stanford University",
      "year": "2020"
    },
    {
      "degree": "B.A. in Statistics",
      "institution": "Harvard University",
      "year": "2018"
    }
  ],
  "certifications": [
    "Google Cloud Certified Professional Data Engineer"
  ],
  "other_qualifications": "Jane holds a Master of Science in Data Science from Stanford University and a Bachelor of Arts in Statistics from Harvard University. Additionally, she is a Google Cloud Certified Professional Data Engineer.",
  "projects": [
    {
      "name": "Prompt Engineering Tool",
      "description": "A Python application that optimizes prompts for the Gemini API using few-shot and chain-of-thought techniques."
    },
    {
      "name": "Text Summarizer",
      "description": "A model built with PyTorch and transformers to generate concise summaries of articles."
    }
  ],
  "is_valid_resume": true
}
"""

# The main prompt template
RESUME_PROCESSOR_TEMPLATE_OLD = """
You are an expert AI system designed for precise and detailed resume data extraction. Your task is to analyze the provided resume text and extract structured information into a specific JSON format.

### Instructions ###
1.  **Objective**: Extract all structured information from the provided resume into a valid JSON object.
2.  **CRITICAL INSTRUCTION FOR SKILL RATING**: For each skill, you MUST generate a proficiency rating (1-10) and a list of sub-skills.
    * **Rating Scale**:
        * **1-3 (Novice)**: Mentioned only in a skills list.
        * **4-6 (Intermediate)**: Used in a project or job role with a basic description.
        * **7-8 (Advanced)**: Used extensively with specific libraries/frameworks.
        * **9-10 (Expert)**: Implied leadership or deep knowledge.
    * **Sub-skills**: Infer and list relevant, specific sub-skills based on the context. For example, "Python" could have sub-skills like "Data Analysis," "Web Development," or "Scripting." "NLP" could have sub-skills like "LLMs," "Prompt Engineering," and "Gemin."
3.  **JSON Structure**: Return a valid JSON object. All field names must be lowercase.
    * `first_name`, `last_name`, `name`, `email`, `linkedin_url`, `phone`, `summary` (all strings).
    * `skills` (array of objects): `name` (string), `rating` (integer), `description` (string explaining the rating), `sub_skills` (array of strings).
    * `experience` (array of objects): `title`, `company`, `duration` (string), `description`.
    * `total_experience` (object):
        * `years` (string): Calculate total experience from all work entries, using '{current_date}' as the current date for jobs with 'Present' as the end duration. Format as "X years Y months".
        * `description` (string): A brief summary of the candidate's total professional experience.
    * `education` (array of objects): `degree`, `institution`, `year`.
    * `certifications` (array of strings).
    * `other_qualifications` (string).
    * `projects` (array of objects): `name`, `description`.
    * `is_valid_resume` (boolean): `True` if document contains resume data, `False` if not a resume.

### Few-Shot Examples ###

---
<example_start>
Resume Text:
{example_resume_text_1}

JSON Output:
{example_json_output_1}
<example_end>
---
<example_start>
Resume Text:
{example_resume_text_2}

JSON Output:
{example_json_output_2}
<example_end>
---

### Final Task ###
Resume Text:
{resume_text}

JSON Output:
"""

def generate_prompt(resume_text: str, current_date: str):
    """Generate prompt for extracting structured information from resume text"""
    return RESUME_PROCESSOR_TEMPLATE_OLD.format(
        current_date=current_date,
        example_resume_text_1=example_resume_text_1,
        example_json_output_1=example_json_output_1,
        example_resume_text_2=example_resume_text_2,
        example_json_output_2=example_json_output_2,
        resume_text=resume_text
    )

def process_resume(resume_text: str):
    """
    Main function to generate prompt, call Gemini, and process the response.
    """
    try:
        gemini_client = GeminiClient(genaiA.GenerativeModel("gemini-2.5-flash"))
        prompt = generate_prompt(resume_text)

        print("==========================")
        print("Generated Prompt:\n", prompt)
        print("==========================")
        
        # Call the Gemini model
        content = gemini_client.call_gemini(prompt, domain="resume_processing")

        # Extract and validate JSON
        cleaned_content = extract_and_validate_json(content)
        
        if not cleaned_content:
            print("Invalid JSON content extracted from resume")
            return None
        
        json_content = json.loads(cleaned_content)
        print("json_content==========")
        print(json.dumps(json_content, indent=2))
        return json_content

    except Exception as e:
        print(f"An error occurred: {e}")
        return None

```

```python
import json
import os

# file_path = "/home/dmns/Desktop/project/other/postgres & mongo/Nishan Sangeeth CV .pdf"

def load_json_content(filename="resume_data.json"):
    """Load JSON content from a file with error handling"""
    try:
        # Check if file exists
        if not os.path.exists(filename):
            print(f"File {filename} does not exist")
            return None
            
        with open(filename, 'r', encoding='utf-8') as f:
            content = json.load(f)
            print(f"Successfully loaded JSON from {filename}")
            return content
            
    except json.JSONDecodeError as e:
        print(f"Invalid JSON format in {filename}: {e}")
        return None
    except Exception as e:
        print(f"Error loading JSON file {filename}: {e}")
        return None

# Usage
json_file = "resume_data.json"
loaded_json_content = load_json_content(json_file)

if loaded_json_content:
    print("Loaded content:", loaded_json_content)
else:
    print("Failed to load JSON content")
```

```python
def transform_resume_data(data: dict, file_path: str = None, user_id: str = None):
        """Transform extracted JSON data into standardized resume format"""
        try:
            # Handle projects with None descriptions
            projects = []
            for project in data.get("projects", []):
                projects.append({
                    "name": project.get("name"),
                    "description": project.get("description") or ""  # Convert None to empty string
                })

            resume_data = {
                "user_id": user_id,
                "first_name": data.get("first_name"),
                "last_name": data.get("last_name"),
                "name": data.get("name", ""),
                "email": data.get("email", ""),
                "phone": data.get("phone", ""),
                "linkedin_url": data.get("linkedin_url"),
                "summary": data.get("summary", ""),
                "skills": data.get("skills", []),
                "experience": data.get("experience", []),
                "total_experience": data.get("total_experience", {}),
                "education": data.get("education", []),
                "certifications": data.get("certifications", []),
                "projects": projects,  # Use the transformed projects list
                "industry": data.get("industry", []),
                "file_path": file_path,
                "is_valid_resume": data.get("is_valid_resume", True)
            }
            return resume_data
        except Exception as e:
            print("Error transforming resume data: %s", str(e))
            return {}
```

```python
def add_embeddings_to_resume(resume_data: dict, embeddings: dict) -> dict:
    """
    Add embeddings to the resume data structure according to the specified format
    
    Returns:
        Updated resume data with embeddings for each section
    """
    try:
        # Create a deep copy to avoid modifying the original
        updated_resume = resume_data.copy()
        
        # Add embeddings to summary section
        if 'summary' in embeddings.get('sections', {}):
            updated_resume['summary'] = {
                "text": resume_data.get('summary', ''),
                "embedding": embeddings['sections']['summary']['embedding']
            }
        
        # Add embeddings to skills section
        if 'skills' in embeddings.get('sections', {}):
            updated_resume['skills'] = {
                "text": resume_data.get('skills', []),
                "embedding": embeddings['sections']['skills']['embedding']
            }
        
        # Add embeddings to experience section
        if 'experience' in embeddings.get('sections', {}):
            experience_with_embeddings = []
            for i, exp in enumerate(resume_data.get('experience', [])):
                exp_copy = exp.copy()
                # Add the experience embedding to each experience entry
                exp_copy['embedding'] = embeddings['sections']['experience']['embedding']
                experience_with_embeddings.append(exp_copy)
            updated_resume['experience'] = experience_with_embeddings
        
        # Add embeddings to education section
        if 'education' in embeddings.get('sections', {}):
            education_with_embeddings = []
            for i, edu in enumerate(resume_data.get('education', [])):
                edu_copy = edu.copy()
                # Add the education embedding to each education entry
                edu_copy['embedding'] = embeddings['sections']['education']['embedding']
                education_with_embeddings.append(edu_copy)
            updated_resume['education'] = education_with_embeddings
            
        if 'total_experience' in embeddings.get('sections', {}):
            # Preserve the original total_experience structure and add embedding
            original_total_exp = resume_data.get('total_experience', {})
            updated_resume['total_experience'] = {
                "years": original_total_exp.get('years', ''),
                "description": original_total_exp.get('description', ''),
                "embedding": embeddings['sections']['total_experience']['embedding']
            }
        
        # Add embeddings to certifications section
        if 'certifications' in embeddings.get('sections', {}):
            updated_resume['certifications'] = {
                "text": resume_data.get('certifications', []),
                "embedding": embeddings['sections']['certifications']['embedding']
            }
        
        # Add embeddings to projects section
        if 'projects' in embeddings.get('sections', {}):
            projects_with_embeddings = []
            for project in resume_data.get('projects', []):
                project_copy = project.copy()
                # Add the projects embedding to each project entry
                project_copy['embedding'] = embeddings['sections']['projects']['embedding']
                projects_with_embeddings.append(project_copy)
            updated_resume['projects'] = projects_with_embeddings
            
        # Add combined embedding at the root level
        if embeddings.get('combined_embedding'):
            updated_resume['combined_embedding'] = embeddings['combined_embedding']
        
        return updated_resume
        
    except Exception as e:
        print(f"Error adding embeddings to resume: {e}")
        return resume_data
```

```python
import json
import os


def load_json_content(filename="jd_data.json"):
    """Load JSON content from a file with error handling"""
    try:
        # Check if file exists
        if not os.path.exists(filename):
            print(f"File {filename} does not exist")
            return None
            
        with open(filename, 'r', encoding='utf-8') as f:
            content = json.load(f)
            print(f"Successfully loaded JSON from {filename}")
            return content
            
    except json.JSONDecodeError as e:
        print(f"Invalid JSON format in {filename}: {e}")
        return None
    except Exception as e:
        print(f"Error loading JSON file {filename}: {e}")
        return None

# Usage
json_file = "jd_data.json"
loaded_json_content = load_json_content(json_file)

if loaded_json_content:
    print("Loaded content:", loaded_json_content)
else:
    print("Failed to load JSON content")


# import json
# import os
# def load_json_content(filename="jd_data.json"):
#     """Load JSON content from a file with error handling"""
#     try:
#         # Check if file exists
#         if not os.path.exists(filename):
#             print(f"File {filename} does not exist")
#             return None
            
#         with open(filename, 'r', encoding='utf-8') as f:
#             content = json.load(f)
#             print(f"Successfully loaded JSON from {filename}")
#             return content
            
#     except json.JSONDecodeError as e:
#         print(f"Invalid JSON format in {filename}: {e}")
#         return None
#     except Exception as e:
#         print(f"Error loading JSON file {filename}: {e}")
#         return None
```

```python
from typing import Any, Dict


def prepare_jd_for_insertion(  jd_data: Dict[str, Any],
                               jd_embedding: Dict[str, Any]
                               ) -> Dict[str, Any]:
        """Prepare JD data with embeddings for database insertion"""
        try:
            # Create a copy of the original JD data
            updated_jd = jd_data.copy()
            
            # Add section embeddings
            sections_embeddings = jd_embedding.get("sections", {})
            
            compulsory_skills = jd_data.get("compulsory_skills", [])
            nice_to_have_skills = jd_data.get("nice_to_have_skills", [])
            optional_skills = jd_data.get("optional_skills", [])
            
            # Combine all skills into a single string
            all_skills = compulsory_skills + nice_to_have_skills + optional_skills
            combined_skills_text = ", ".join(all_skills)
            
            # Store as a list with one element (if needed for consistency)
            # or just as a string depending on your database requirements
            combined_skills_for_storage = [combined_skills_text] 
            
            if 'preferred_industry' in sections_embeddings:
                updated_jd['preferred_industry'] = {
                    "text": jd_data.get('preferred_industry', []),
                    "embedding": jd_embedding['sections']['preferred_industry']['embedding']
                }
            
            if 'description' in sections_embeddings:
                updated_jd['description'] = {
                    "text": jd_data.get('description', ''),
                    "embedding": jd_embedding['sections']['description']['embedding']
                }
            
            if 'compulsory_skills' in sections_embeddings:
                updated_jd['compulsory_skills'] = {
                    "text": jd_data.get('compulsory_skills', []),
                    "embedding": jd_embedding['sections']['compulsory_skills']['embedding']
                }
            
            if 'nice_to_have_skills' in sections_embeddings:
                updated_jd['nice_to_have_skills'] = {
                    "text": jd_data.get('nice_to_have_skills', []),
                    "embedding": jd_embedding['sections']['nice_to_have_skills']['embedding']
                }
                
            if 'optional_skills' in sections_embeddings:
                updated_jd['optional_skills'] = {
                    "text": jd_data.get('optional_skills', []),
                    "embedding": jd_embedding['sections']['optional_skills']['embedding']
                }
                
            if 'combined_skills' in sections_embeddings:
                updated_jd['combined_skills'] = {
                    "text": combined_skills_for_storage,
                    "embedding": jd_embedding['sections']['combined_skills']['embedding']
                }
                
            if 'preferred_experience' in sections_embeddings:
                updated_jd['preferred_experience'] = {
                    "text": jd_data.get('preferred_experience', ''),
                    "embedding": jd_embedding['sections']['preferred_experience']['embedding']
                }
            if 'other_qualifications' in sections_embeddings:
                updated_jd['other_qualifications'] = {
                    "text": jd_data.get('other_qualifications', ''),
                    "embedding": jd_embedding['sections']['other_qualifications']['embedding']
                }
            
            if jd_embedding.get('combined_embedding'):
                updated_jd['combined_embedding'] = jd_embedding['combined_embedding']
        
            return updated_jd
            
        except Exception as e:
            print(f"Error preparing JD for insertion: {e}")
            return jd_data
```

```python
import psycopg2

def create_and_populate_skills_table():
    """Create skills table and insert all skills data in one function"""
    conn = None
    try:
        # Connect to PostgreSQL
        conn = psycopg2.connect(
            host="localhost",
            database="resume_db",
            user="dmns",
            password="dmns",
            port=5432
        )
        
        with conn.cursor() as cursor:
            # Create table
            cursor.execute("""
                CREATE TABLE IF NOT EXISTS skills (
                    skill_id SERIAL PRIMARY KEY,
                    skill VARCHAR(100) UNIQUE NOT NULL,
                    created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
                )
            """)
            
            # Insert all skills
            skills = [
                "Python", "Machine Learning", "Deep Learning", "Natural Language Processing",
                "Computer Vision", "Statistical Analysis", "Neural Networks", "Model Deployment",
                "Django", "Flask", "RESTful APIs", "SQL", "NoSQL", "Data Structures",
                "Testing", "Debugging", "Git", "React", "Node", "Docker", "AWS", "Azure",
                "GCP", "CI/CD Pipelines", "Spark", "Hadoop", "Agile Methodology",
                "Problem-Solving Skills", "Java", "HTML", "CSS", "JavaScript", "Microservices",
                "LangChain", "RAG", "ONNX", "TensorFlow", "ComfyUI", "Llama", "CNN",
                "Pandas", "Numpy", "OpenCV"
            ]
            
            for skill in skills:
                cursor.execute(
                    "INSERT INTO skills (skill) VALUES (%s) ON CONFLICT (skill) DO NOTHING",
                    (skill,)
                )
            
            # Count inserted skills
            cursor.execute("SELECT COUNT(*) FROM skills")
            count = cursor.fetchone()[0]
            
            conn.commit()
            print(f"✓ Success! Created skills table with {count} skills")
            
    except Exception as e:
        print(f"Error: {e}")
        if conn:
            conn.rollback()
    finally:
        if conn:
            conn.close()

# Run the function
create_and_populate_skills_table()
```
