```python
def search_resumes_by_jd_compulsory_skills(jd_id):
    """
    Search for resumes that match the compulsory skills of a specific JD
    """
    try:
        # Connect to both databases
        jd_conn = psycopg2.connect(
            host="localhost",
            database="jd_db",
            user="dmns",
            password="dmns",
            port=5432
        )
        
        resume_conn = psycopg2.connect(
            host="localhost",
            database="resume_db",
            user="dmns",
            password="dmns",
            port=5432
        )

        all_skills = []
        
        # Step 1: Get compulsory skills from the JD
        with jd_conn.cursor() as cursor:
            # Get compulsory skills
            query = """SELECT skill FROM jd_compulsory_skills WHERE jd_id = %s;"""
            cursor.execute(query, (jd_id,))
            compulsory_skills = [skill[0] for skill in cursor.fetchall()]
            all_skills.extend(compulsory_skills)
            print(f"Compulsory skills: {compulsory_skills}")
            
            # Get nice to have skills
            query = """SELECT skill FROM jd_nice_to_have_skills WHERE jd_id = %s;"""
            cursor.execute(query, (jd_id,))
            nice_to_have_skills = [skill[0] for skill in cursor.fetchall()]
            all_skills.extend(nice_to_have_skills)
            print(f"Nice to have skills: {nice_to_have_skills}")
            
            # Get optional skills
            query = """SELECT skill FROM jd_optional_skills WHERE jd_id = %s;"""
            cursor.execute(query, (jd_id,))
            optional_skills = [skill[0] for skill in cursor.fetchall()]
            all_skills.extend(optional_skills)
            print(f"Optional skills: {optional_skills}")
            
            if not all_skills:
                print(f"No skills found for JD ID: {jd_id}")
                return []
            
            print(f"All skills for JD {jd_id}: {all_skills}")
            return all_skills
            
    except psycopg2.Error as e:
        print(f"Database error: {e}")
        return []
    except ValueError:
        print("Please enter a valid JD ID (number).")
        return []
    except KeyboardInterrupt:
        print("\nOperation cancelled by user.")
        return []
    
    
# If you're still getting nested lists, use this function to flatten them:
def flatten_skills_list(skills_list):
    """
    Flatten a nested list of skills into a single flat list
    """
    flat_list = []
    for item in skills_list:
        if isinstance(item, list):
            flat_list.extend(flatten_skills_list(item))
        else:
            flat_list.append(item)
    return flat_list

# Usage:
```

```python
def get_all_skills_flat(jd_id):
    """
    Get all skills as a flat list from all skill tables
    """
    try:
        jd_conn = psycopg2.connect(
            host="localhost",
            database="jd_db",
            user="dmns",
            password="dmns",
            port=5432
        )

        with jd_conn.cursor() as cursor:
            # Get all skills from all tables in one go
            query = """
                SELECT skill FROM jd_compulsory_skills WHERE jd_id = %s
                UNION ALL
                SELECT skill FROM jd_nice_to_have_skills WHERE jd_id = %s
                UNION ALL
                SELECT skill FROM jd_optional_skills WHERE jd_id = %s
            """
            cursor.execute(query, (jd_id, jd_id, jd_id))
            
            # Get all skills as a flat list
            skills = [skill[0] for skill in cursor.fetchall() if skill[0]]
            
            flat_skills = []
            for sublist in skills:
                if isinstance(sublist, list):
                    flat_skills.extend(sublist)
                else:
                    flat_skills.append(sublist)

            print(f"Flat skills: {flat_skills}")
            
            print(f"All skills for JD {jd_id}: {skills}")
            return flat_skills
            
    except psycopg2.Error as e:
        print(f"Database error: {e}")
        return []
```

```python
from psycopg2.extras import Json


def insert_jd_skills(jd_id, skills_list):
    """
    Insert skills into jd_skills table with embeddings
    """
    try:
        jd_conn = psycopg2.connect(
            host="localhost",
            database="jd_db",
            user="dmns",
            password="dmns",
            port=5432
        )

        with jd_conn.cursor() as cursor:
            # Convert skills list to string for embedding generation
            skills_string = ', '.join(skills_list)
            
            # Generate embedding
            from embeddings import EmbeddingGenerator
            generator = EmbeddingGenerator()
            combined_embedding = generator.generate_embedding(skills_string)
            
            # Insert into jd_skills table
            # Since skill can be JSONB, we'll store both the list and string representation
            cursor.execute("""
                INSERT INTO jd_skills (jd_id, skill, embedding)
                VALUES (%s, %s, %s)
            """, (
                jd_id,
                Json(skills_list),  # Store the original list directly
                combined_embedding  # Embedding generated from the string
            ))
            
            jd_conn.commit()
            print(f"Successfully inserted skills for JD {jd_id}")
            
    except psycopg2.Error as e:
        print(f"Database error: {e}")
        jd_conn.rollback()
    except Exception as e:
        print(f"Error generating embedding: {e}")
        jd_conn.rollback()
```

```python
import psycopg2
import numpy as np
from pgvector.psycopg2 import register_vector

def search_resumes_by_jd_skills_vector(jd_id, limit=20, similarity_threshold=0.5):
    """
    Search for resumes that match the compulsory skills of a specific JD using vector similarity
    Returns: (results_list, resume_ids_list) - both lists of matching resumes
    """
    jd_conn = None
    resume_conn = None
    try:
        jd_conn = psycopg2.connect(
            host="localhost",
            database="jd_db",
            user="dmns",
            password="dmns",
            port=5432
        )
        
        resume_conn = psycopg2.connect(
            host="localhost",
            database="resume_db",
            user="dmns",
            password="dmns",
            port=5432
        )
        
        register_vector(resume_conn)
        
        with jd_conn.cursor() as cursor:
            query = """
            SELECT skill, embedding
            FROM jd_skills
            WHERE jd_id = %s AND embedding IS NOT NULL;
            """
            cursor.execute(query, (jd_id,))
            jd_skills = cursor.fetchall()
            print("jd_skills", jd_skills)
            
            if not jd_skills:
                print(f"No compulsory skills found for JD ID: {jd_id}")
                return [], []
        
        print(f"Found {len(jd_skills)} compulsory skills for JD {jd_id}")
        
        all_matches = []
        
        for skill_name, jd_embedding in jd_skills:
            with resume_conn.cursor() as cursor:
                resume_query = """
                SELECT rs.resume_id, rs.skill, rs.embedding,
                       1 - (rs.embedding <=> %s) as cosine_similarity
                FROM resume_skills rs
                WHERE 1 - (rs.embedding <=> %s) > %s
                ORDER BY cosine_similarity DESC
                """
                
                cursor.execute(resume_query, (jd_embedding, jd_embedding, similarity_threshold))
                matches = cursor.fetchall()
                all_matches.extend(matches)
                print(f"Skill '{skill_name}': Found {len(matches)} matches")
        
        
        # Process the matches to aggregate by resume_id
        resume_results = {}
        
        for match in all_matches:
            resume_id, skill, embedding, similarity = match
            if resume_id not in resume_results:
                resume_results[resume_id] = {
                    'resume_id': resume_id,
                    'matching_skills': [],
                    'total_similarity': 0,
                    'skill_count': 0
                }

            resume_results[resume_id]['matching_skills'].append({
                'skill': skill,
                'similarity': similarity
            })
            resume_results[resume_id]['total_similarity'] += similarity
            resume_results[resume_id]['skill_count'] += 1

        # Convert to list and calculate average similarity
        results_list = []
        for resume_id, data in resume_results.items():
            avg_similarity = data['total_similarity'] / data['skill_count'] if data['skill_count'] > 0 else 0
            results_list.append({
                'resume_id': resume_id,
                'matching_skills': data['matching_skills'],
                'avg_similarity': avg_similarity,
                'matching_skills_count': data['skill_count']
            })

        # Sort by average similarity and number of matching skills
        results_list.sort(key=lambda x: (x['avg_similarity'], x['matching_skills_count']), reverse=True)

        print(f"\nFound {len(results_list)} matching resumes:")
        for result in results_list[:min(25, len(results_list))]:
            print(f"Resume {result['resume_id']}: Avg similarity={result['avg_similarity']:.3f}, Skills matched={result['matching_skills_count']}")
        
        # Get all selected resume IDs
        selected_results = results_list[:limit]
        resume_ids_list = [result['resume_id'] for result in selected_results]
        
        return selected_results, resume_ids_list
        
    except Exception as e:
        print(f"Error: {e}")
        return [], []
    finally:
        if jd_conn:
            jd_conn.close()
        if resume_conn:
            resume_conn.close()

# Usage
```

```python
import psycopg2
import numpy as np
from pgvector.psycopg2 import register_vector

def search_resumes_by_jd_compulsory_skills_vector(jd_id, resume_ids_list, limit=20, similarity_threshold=0.5):
    """
    Search for resumes that match the compulsory skills of a specific JD using vector similarity
    Only searches within the provided list of resume IDs
    Returns: (results_list, filtered_resume_ids_list) - both lists of matching resumes
    """
    jd_conn = None
    resume_conn = None
    try:
        jd_conn = psycopg2.connect(
            host="localhost",
            database="jd_db",
            user="dmns",
            password="dmns",
            port=5432
        )
        
        resume_conn = psycopg2.connect(
            host="localhost",
            database="resume_db",
            user="dmns",
            password="dmns",
            port=5432
        )
        
        register_vector(resume_conn)
        
        with jd_conn.cursor() as cursor:
            query = """
            SELECT skill, embedding
            FROM jd_compulsory_skills
            WHERE jd_id = %s AND embedding IS NOT NULL;
            """
            cursor.execute(query, (jd_id,))
            jd_compulsory_skills = cursor.fetchall()
            print("jd_compulsory_skills", jd_compulsory_skills)
            
            if not jd_compulsory_skills:
                print(f"No compulsory skills found for JD ID: {jd_id}")
                return [], []
        
        print(f"Found {len(jd_compulsory_skills)} compulsory skills for JD {jd_id}")
        
        # If no resume IDs provided, return empty results
        if not resume_ids_list:
            print("No resume IDs provided for filtering")
            return [], []
        
        all_matches = []
        
        for skill_name, jd_embedding in jd_compulsory_skills:
            with resume_conn.cursor() as cursor:
                # Modified query to filter by resume_ids_list
                resume_query = """
                SELECT rs.resume_id, rs.skill, rs.embedding,
                       1 - (rs.embedding <=> %s) as cosine_similarity
                FROM resume_skills rs
                WHERE 1 - (rs.embedding <=> %s) > %s
                AND rs.resume_id = ANY(%s)
                ORDER BY cosine_similarity DESC
                """
                
                cursor.execute(resume_query, (jd_embedding, jd_embedding, similarity_threshold, resume_ids_list))
                matches = cursor.fetchall()
                all_matches.extend(matches)
                print(f"Skill '{skill_name}': Found {len(matches)} matches in filtered resumes")
        
        
        # Process the matches to aggregate by resume_id
        resume_results = {}
        
        for match in all_matches:
            resume_id, skill, embedding, similarity = match
            if resume_id not in resume_results:
                resume_results[resume_id] = {
                    'resume_id': resume_id,
                    'matching_skills': [],
                    'total_similarity': 0,
                    'skill_count': 0
                }

            resume_results[resume_id]['matching_skills'].append({
                'skill': skill,
                'similarity': similarity
            })
            resume_results[resume_id]['total_similarity'] += similarity
            resume_results[resume_id]['skill_count'] += 1

        # Convert to list and calculate average similarity
        results_list = []
        for resume_id, data in resume_results.items():
            avg_similarity = data['total_similarity'] / data['skill_count'] if data['skill_count'] > 0 else 0
            results_list.append({
                'resume_id': resume_id,
                'matching_skills': data['matching_skills'],
                'avg_similarity': avg_similarity,
                'matching_skills_count': data['skill_count']
            })

        results_list.sort(key=lambda x: (x['avg_similarity'], x['matching_skills_count']), reverse=True)

        print(f"\nFound {len(results_list)} matching resumes in filtered list:")
        for result in results_list[:min(25, len(results_list))]:
            print(f"Resume {result['resume_id']}: Avg similarity={result['avg_similarity']:.3f}, Skills matched={result['matching_skills_count']}")

        selected_results = results_list[:limit]
        filtered_resume_ids_list = [result['resume_id'] for result in selected_results]
        
        return selected_results, filtered_resume_ids_list
        
    except Exception as e:
        print(f"Error: {e}")
        return [], []
    finally:
        if jd_conn:
            jd_conn.close()
        if resume_conn:
            resume_conn.close()

```
